```python
import pandas as pd
```

```python
r1 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Dominos", "Pizza Hut", "KFC", "Subway"],
    "city_id": [1, 1, 2, 3],
    "cost": [300, 400, 350, 250]
})
r1
```

```python
merged = pd.merge(all_r, cities, on="city_id", how="left")
merged = pd.merge(merged, ratings, on="id", how="left")
merged
```

```python
merged["city"] = merged["city"].fillna("Unknown")
merged["state"] = merged["state"].fillna("Unknown")
merged["rating"] = merged["rating"].fillna(merged["rating"].mean())
merged
```

```python
merged.sort_values("rating", ascending=False)
```

```python
order_totals = orders.groupby("id")["amount"].sum().reset_index()
order_totals
```

```python
final = pd.merge(merged, order_totals, on="id", how="left")
final["amount"] = final["amount"].fillna(0)
final
```

```python
final.groupby("state")[["cost", "rating", "amount"]].mean()
```
