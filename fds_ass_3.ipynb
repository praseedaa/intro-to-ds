```python
csk_df = pd.DataFrame({
    "Player": [
        "MS Dhoni",
        "Ruturaj Gaikwad",
        "Devon Conway",
        "Ravindra Jadeja",
        "Shivam Dube",
        "Moeen Ali",
        "Deepak Chahar",
        "Tushar Deshpande",
        "Maheesh Theekshana",
        "Matheesha Pathirana"
    ],
    "Age": [
        41, 26, np.nan, 34, 29,
        36, 30, np.nan, 22, 20
    ],
    "Runs": [
        104, 590, 672, np.nan, 418,
        124, 0, 21, np.nan, 3
    ],
    "Role": [
        "Wicketkeeper",
        "Batsman",
        "Batsman",
        "All-rounder",
        "All-rounder",
        "All-rounder",
        "Bowler",
        "Bowler",
        "Bowler",
        np.nan
    ]
})
csk_df["Runs"] = csk_df["Runs"].fillna(
    csk_df["Runs"].mean()
)
csk_df["Age"] = csk_df["Age"].fillna(
    csk_df["Age"].median()
)
csk_df["Role"] = csk_df["Role"].fillna(
    csk_df["Role"].mode()[0]
)
print(csk_df.head(5))
```

```python
from scipy.stats import zscore
sample = retail[
    retail["UnitPrice"] > 0
].sample(
    1000,
    random_state=42
).copy()
sample["UnitPrice_Z"] = zscore(
    sample["UnitPrice"]
)
print(
    sample[
        ["UnitPrice", "UnitPrice_Z"]
    ].head()
)
```

```python
sample_clean = sample.copy()
median_price = sample_clean["UnitPrice"].median()
sample_clean.loc[
    (sample_clean["UnitPrice_Z"] > 3) |
    (sample_clean["UnitPrice_Z"] < -3),
    "UnitPrice"
] = median_price
print(
    sample_clean[
        ["UnitPrice", "UnitPrice_Z"]
    ].head()
)
```

```python
from sklearn.preprocessing import MinMaxScaler
student_df = pd.DataFrame({
    "Name": [
        "Arjun",
        "Rohan",
        "Vikram",
        "Karthik",
        "Aditya",
        "Nikhil",
        "Siddharth"
    ],
    "Marks": [85, 78, 92, 74, 88, 81, 95],
    "Attendance": [90, 85, 95, 80, 88, 92, 96]
})
print("Before Scaling:")
print(student_df)
scaler = MinMaxScaler()
student_df[
    ["Marks", "Attendance"]
] = scaler.fit_transform(
    student_df[
        ["Marks", "Attendance"]
    ]
)
print("After Min-Max Scaling:")
print(student_df)
```

```python
from sklearn.preprocessing import StandardScaler
student_df = pd.DataFrame({
    "Name": [
        "Arjun",
        "Rohan",
        "Vikram",
        "Karthik",
        "Aditya",
        "Nikhil",
        "Siddharth"
    ],
    "Marks": [85, 78, 92, 74, 88, 81, 95]
})
scaler = StandardScaler()
student_df["Marks_Standardized"] = (
    scaler.fit_transform(
        student_df[["Marks"]]
    )
)
print(student_df)
```

```python
import pandas as pd
emp1 = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105],
    'Name': ['Meera', 'Kavya', 'Rahul', 'Varun', 'Divya'],
    'Dept': ['IT', 'HR', 'Finance', 'IT', 'Marketing']
})
emp2 = pd.DataFrame({
    'Emp_ID': [106, 107, 108, 109, 110],
    'Name': ['Sanjay', 'Harsha', 'Manoj', 'Kiran', 'Pooja'],
    'Dept': ['HR', 'IT', 'Finance', 'Marketing', 'IT']
})
salary = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Salary': [45000, 40000, 55000, 50000, 42000,
               48000, 46000, 58000, 44000, 52000]
})
emp = pd.concat([emp1, emp2], ignore_index=True)
print("Combined Employee Details:")
print(emp)
```

```python
result = pd.merge(emp, salary, on='Emp_ID')
print("\nAfter Merging Salary:")
print(result)
```
