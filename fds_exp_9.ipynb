# FDS Experiment 9: Iris Dataset EDA and Visualization

Run each cell in order. This notebook covers plot types, figures and subplots, colors, markers, line styles, labels, legends, saving plots, distribution plots, and exploratory data analysis.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load the Iris dataset

```python
df = sns.load_dataset('iris')
df.head()
```

## 3. Inspect the dataset

```python
print(df.shape)
print(df.columns)
df.info()
print(df.isnull().sum())
```

## 4. Line plot

Use a line plot to show values across an ordered sequence of observations.

```python
plt.plot(df.index, df['sepal_length'])
plt.title('Sepal Length')
plt.xlabel('Observation')
plt.ylabel('Sepal Length')
plt.show()
```

## 5. Bar plot

Use a bar plot to compare categories.

```python
avg = df.groupby('species')['petal_length'].mean()

plt.bar(avg.index, avg.values)
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.show()
```

## 6. Histogram

Use a histogram to show the frequency distribution of numerical data.

```python
plt.hist(df['sepal_length'], bins=10, edgecolor='black')
plt.title('Sepal Length Distribution')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

## 7. Pie chart

Use a pie chart to show proportions of categories.

```python
counts = df['species'].value_counts()

plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%')
plt.title('Species Proportion')
plt.show()
```

## 8. Scatter plot

Use a scatter plot to examine the relationship between two numerical variables.

```python
plt.scatter(df['petal_length'], df['petal_width'])
plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.show()
```

## 9. Colors, markers, and line styles

```python
plt.plot(df.index[:20], df['sepal_length'][:20],
         color='red', marker='o', linestyle='--')
plt.title('Sepal Length')
plt.xlabel('Observation')
plt.ylabel('Sepal Length')
plt.show()
```

## 10. Labels and legends

```python
plt.plot(df.index[:10], df['sepal_length'][:10],
         label='Sepal Length', color='blue')
plt.plot(df.index[:10], df['sepal_width'][:10],
         label='Sepal Width', color='red')
plt.xlabel('Observation')
plt.ylabel('Measurement')
plt.title('Sepal Measurements')
plt.legend()
plt.show()
```

## 11. Figures and subplots

A figure is the overall plotting area. Subplots are individual plots inside a figure.

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].hist(df['sepal_length'])
ax[0].set_title('Sepal Length')

ax[1].scatter(df['petal_length'], df['petal_width'])
ax[1].set_title('Petal Measurements')

plt.tight_layout()
plt.show()
```

## 12. Save a plot to a file

The plot is saved as `iris_plot.png` in the current working directory.

```python
plt.plot(df['sepal_length'])
plt.title('Sepal Length')
plt.savefig('iris_plot.png', bbox_inches='tight')
plt.show()
```

## 13. Distribution plot with KDE

```python
sns.histplot(df['petal_length'], kde=True)
plt.title('Petal Length Distribution')
plt.show()
```

## 14. Box plot

A box plot shows the median, quartiles, spread, and potential outliers.

```python
sns.boxplot(data=df, x='species', y='petal_length')
plt.title('Petal Length by Species')
plt.show()
```

## 15. Violin plot

A violin plot shows the shape of a distribution and its spread.

```python
sns.violinplot(data=df, x='species', y='petal_length')
plt.title('Petal Length Distribution by Species')
plt.show()
```

## 16. Exploratory Data Analysis: summary statistics

```python
print(df.describe())
```

## 17. Species counts

```python
print(df['species'].value_counts())
```

## 18. Mean measurements by species

```python
print(df.groupby('species').mean(numeric_only=True))
```

## 19. Correlation heatmap

Correlation values range from -1 to +1. Correlation does not establish causation.

```python
corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Feature Correlation')
plt.show()
```

## 20. Pairplot

This compares numerical features against each other, with points colored by species.

```python
sns.pairplot(df, hue='species')
plt.show()
```

## 21. EDA observations

- The dataset contains three species with 50 observations each.
- Petal length and petal width are useful features for distinguishing species.
- Setosa tends to have smaller petals than the other two species.
- Versicolor and virginica overlap more in some measurements.
- Scatter plots reveal clusters, while box plots help compare medians and spreads.

These are descriptive findings, not proof that a model will classify every flower correctly.
